##문제:

n개의 정수로 이루어진 임의의 수열이 주어진다. 우리는 이 중 연속된 몇 개의 수를 선택해서 구할 수 있는 합 중 가장 큰 합을 구하려고 한다. 단, 수는 한 개 이상 선택해야 한다. 또, 수열에서 수를 하나 제거할 수 있다. (제거하지 않아도 된다)

예를 들어서 10, -4, 3, 1, 5, 6, -35, 12, 21, -1 이라는 수열이 주어졌다고 하자. 여기서 수를 제거하지 않았을 때의 정답은 12+21인 33이 정답이 된다.

만약, -35를 제거한다면, 수열은 10, -4, 3, 1, 5, 6, 12, 21, -1이 되고, 여기서 정답은 10-4+3+1+5+6+12+21인 54가 된다.

##입력:

첫째 줄에 정수 n(1 ≤ n ≤ 100,000)이 주어지고 둘째 줄에는 n개의 정수로 이루어진 수열이 주어진다. 수는 -1,000보다 크거나 같고, 1,000보다 작거나 같은 정수이다.

##출력:

첫째 줄에 답을 출력한다.

## 문제 접근

이 문제가 먼저 DP(Dynamic Programing)로 풀 수 있는 문제인지 고려해 보아야 한다.

  만약 0\~i번째 까지의 숫자를 고려할 때 i번째 숫자가 연속된 수를 선택할 때 포함된다고 가정하면, 다음 0\~i+1번째 숫자를 고려할 시에는 i+1번째 숫자를 그 0\~i번째 숫자중 연속된 숫자의 합에 추가로 더해보아 더 커지면 그 새롭게 합해준 결과값이 0\~i+1번째 숫자를 고려할 시 연속 숫자합의 최대값이 된다. 따라서 하위문제를 통해 상위문제를 해결할 수 있을 듯 하다.

  다음으로 이 문제를 어떻게 배열화 하여야 memoization을 활용할 수 있을지, 어떻게 자연스럽게 하위문제와 상위문제를 연결시켜 점화식을 세울 수 있을지 고민해 보아야 한다.

  먼저 D[i]라는 배열을 만들어 이를 '0\~i까지 길이에서 연속으로 숫자를 선택하여 구할 수 있는 최대합' 이라고 정의해 볼 수 있는데, 이렇게 정의하면 문제가 생길 수 있다. 바로 위 예시 입력을 통해 생각해보면 D[0],D[1],D[2],D[3]이 모두 10이 되는데 이는 하위문제의 풀이를 통해 상위문제를 풀기에는 적절하지 못한 형태이다. 이는 그 다음 숫자를 더해줄 때도 연속인지 아닌지를 판단하기에도 방해가 되기 때문이다.(만약 D[6]이 0\~3번째 숫자의 합이었다면 연속성이 끊겨 D[7]을 구할 시 D[6]의 값을 사용할 수 없다)

  따라서 점화식을 만들기 위해서는 배열 D[i]를 '0\~i까지 길이에서 i번째 숫자를 포함하여 연속으로 숫자를 선택하여 구할 수 있는 최대합' 으로 정의하는 것이 더 적절하다.(이렇게 정의하면 D[i+1]을 구하기 위해서 D[i]를 활용할 수 있게 됨)

  이제 문제의 조건처럼 숫자 하나를 제거할 수 있는 선택을 어떻게 반영해야할지를 고민해야 한다. 이에 대한 답은 D[i]를 i번째 숫자를 기준으로 좌우로 동시에 구하는 것이다. 점화식 L[i]를 기존의 D[i]처럼 '0\~i까지 길이에서 i번째 숫자를 포함하여 연속으로 숫자를 선택하여 구할 수 있는 최대합'이라고 하고, R[i]를 'i\~N(마지막)까지 길이에서 i번째 숫자를 포함하여 연속으로 숫자를 선택하여 구할 수 있는 최대합' 으로 정의한다면, L[i-1]+R[i+1]과 L[i](또는 R[i])를 비교하여 더 커지는 값을 정답으로 하면 숫자를 하나 제거할 수 있다는 선택을 반영할 수 있게 된다.

  이제 위에 설명을 바탕으로 점화식을 세워보자.여기서 문제에서 제시하는 입력 배열을 A[i]라고 하자.

  $L[i] = max(A[i],L[i-1]+A[i])$

  $R[i] = max(A[i],R[i+1]+A[i])$

  마지막으로 초기조건은 다음과 같다.(여기서 N은 입력받는 배열의 길이)

  $L[0] = A[0]$

  $R[N-1] = A[N-1]$

## 코드 작성

```python
## 입력받기

N = int(input())
A = list(map(int, input().split()))

print('='*100)
print('A= ',A)
```

```python
##배열 L[i]와 R[i] 만들기

# 왼쪽부터 i번째 숫자를 포함하는 연속숫자의 최대합 L[i]
L = [0]*N #0으로 일단 모두 초기화
L[0] = A[0] #초기조건

print('='*100)
print('L= ',L)


# 오른쪽부터 i번째 숫자를 포함하는 연속숫자의 최대합 R[i]
R = [0]*N
R[N-1] = A[N-1] #초기조건: 인덱스가 0부터 N-1까지 이므로 N-1번째 숫자가 오른쪽 끝이다.

print('='*100)
print('R= ',R)
```

```python
## 점화식으로 L[i] R[i] 업데이트

#L[i] 업데이트
for i in range(1,N):
  L[i] = max(A[i], L[i-1]+A[i])

#R[i] 업데이트(R의 인덱스는 N-1부터 0까지 역순이므로, 초기조건 N-1을 제외한 N-2부터 -1초과(0이상)까지 역순으로(step=-1))
for i in range(N-2,-1,-1):
  R[i] = max(A[i], R[i+1]+A[i])



print('='*100)
print('L= ',L)

print('='*100)
print('R= ',R)
```

```python
## 마지막 정답 출력

#만약 숫자를 제거하지 않는다면 정답은 L[i]중 최대값이고 제거한다면 L[i-1]+R[i+1]중 최대값이다.
#최종 정답은 위 둘 중 큰 값을 출력해내는 것이다.
result = max(L) #먼저 제거 안했을 때의 정답값으로 result 초기화
#제거했을 때의 정답과 비교하며 더 큰값을 최종 정답으로 함(범위는 L[i-1]+R[i+1]의 계산상 좌우 1씩 여유가 필요하므로 1~N-2)
for i in range(1,N-1):
  result = max(result, L[i-1]+R[i+1])

print(result)
```

```python
#백준 13398번 전체 코드
# import sys
# sys.stdin.readline

N = int(input())
A = list(map(int, input().split()))

L = [0]*N
L[0] = A[0]

R = [0]*N
R[N-1] = A[N-1]

for i in range(1,N):
  L[i] = max(A[i], L[i-1]+A[i])

for i in range(N-2,-1,-1):
  R[i] = max(A[i], R[i+1]+A[i])

result = max(L)

for i in range(1,N-1):
  result = max(result, L[i-1]+R[i+1])

print(result)
```
