## Proof of Optimality

위와 같은 방식(greedy)으로 해를 구하는 것이 전체 구조에 대한 최적해인 이유에 대한 증명을 살펴보자.(wikipedia)

여기서 회의일정을 일종의 활동이라고 하자.(활동 선택 문제)

먼저, 활동 $i$의 시작 시각을 $s_{i}$, 종료 시각을 $f_{i}$ 라고 할때, 두 활동 $i$,$j$가 겹치지 않는다는 말은 $s_{i} \geq f_{j}$ 또는 $s_{j} \geq f_{i}$ 라는 말이다.

또한 어떠한 활동들의 집합을 $A$ 라고 한다면, 해당 집합이 해가 되려면 집합 내 활동들(원소들)은 위의 서로 겹치지 않는다는 조건을 만족하는 활동들이어야 한다.

가장 최적일 때(최대한 많은 겹치지 않는 활동들을 포함)를 의미하는 $A^{*}$는 정의상 어떠한 해집합 $A$를 잡아도 $|A^{*}| \geq |A|$ 이어야 한다.($|A|$는 $A$의 원소의 개수, 즉 활동들 개수를 의미)

이제 증명을 해보자.

전체 주어진 n개의 활동들을 종료시간에 따라 오름차순 정렬한 집합을 $S = \{1,2,3, ... , n\}$ 라고 하자.

$S$의 부분집합 $A^*$가 최적의 해 중 하나라고 가정하자(최적해는 여러개 일수도 있음).($A^*$도 종료시각에 따라 정렬되어 있다) $A^*$에서의 첫 활동을 $k$라고 할때 $k \neq 1$라고 하자.(최적해라고 해서 S의 맨 처음 활동(=1)(가장 먼저 끝나는)을 포함할 필요는 없으므로) 즉, 이 최적해는 greedy 선택으로 시작하지 않는 해인 것이다.

만약, 또다른 해집합 $B = (A^* \backslash \{k\}) \cup \{ 1 \}$라고 할때($(A^* \backslash \{k\}) \cup \{ 1 \}$ 는 집합 $A$에서 원소 $k$를 빼고 원소 $1$을 추가한다는 뜻), 이 해 $B$ 또한 최적해임을 보이려고 한다. 다시말해, S의 가장 먼저 끝나는 활동(S의 첫 원소,1)이 포함되어 있지 않은 최적해가 있을 때 그 최적해의 맨 첫 활동(k)을 $S$의 첫 원소(1)로 대체해도 이 집합 또한 최적해임을 증명하고자 한다.

$f_{1} \leq f_{k}$ 즉 원소 $1$의 종료 시각이 $k$의 종료시각보다 빠르고, $f_{k} \leq s_{i} (i \in A^*)$로써 A의 원소들이 서로소(disjoint)였으므로 1을 k로 대체한 $B$ 또한 겹치지 않는 서로소 조건을 만족한다.

$f_{1} \leq s_{i} (i \in A^* \ )$ -- 식1)

그리고 집합 $A^*$에서 하나의 원소를 제거한 후 다시 원소 하나를 추가한 집합 $B$는 같은 크기를 가진다.

$|A^*| = |B|$ -- 식2)

$B$는 모든 원소들이 서로 겹치지도 않고(식1) 크기도 $A^*$와 같으므로(식2) $B$ 또한 최적해이다.(원소를 교체함으로써 원소 개수가 늘어나지 않으므로)

이제 greedy한 선택을 하여 1을 원소로 하는 최적해를 무조건 만들 수 있다는 것을 확인하였다. 이제 이 greedy choice로 최적해의 첫 원소를 1로 고정하면, 전체 문제는 남은 원소들의 선택(하위 문제)로 축소되며, 이는 최적 부분 구조로 나타난다.

이제 마지막으로 부분 문제에서도 greedy choice가 포함된 최적해가 존재함을 보이면 된다.

$S'$를 $S$의 가장 빨리 끝나는 활동 $1$이 끝난 후 선택할 수 있는 활동들의 집합이라고 하자. 즉, $S' = \{ i \in S | s_{i} \geq f_{1} \}$

$A$를 $S$의 최적해 라고 하고, $A'$를 $A$에서 $1$을 뺀 부분인 $S'$의 해라고 하자. 즉, $A' = A \backslash \{1\}$

귀류법으로 증명하기 위해 $A'$가 $S'$에서의 최적해가 아니라고 가정하자. 즉, $A'$보다 더 많은 활동을 담고 있으며 $S'$의 greedy choice을 포함하는 해 집합 $B'$가 존재한다는 것이다($|B'| > |A'|$). 여기에 $\{1\} \cup B' = B$를 만들면 {1}은 $S'$ 의 모든 활동보다 일찍 끝나므로 $B$는 $S$에서 겹치지 않는 해가 되어 $|B| = 1 + |B'| > 1 + |A'| = |A|$가 되어 $A$보다 최적인 해가 된다. 이는 $A$가 최적이라는 가정에 모순되므로 결국 $A'$는 $S'$에서도 최적이어야 한다.

이 $S'$와 $A'$로 $S$와 $A$로 했던 위 증명을 다시 적용하면 '그리디한 선택을 포함하는 최적해를 찾을 수 있다'가 되고 귀납적으로 진행하면 결국 부분문제 또한 greedy한 choice를 포함하는 최적해를 구할 수 있게 된다.

## 코드 작성

```python
## 입력 받기
from queue import PriorityQueue

N = int(input())

pq = PriorityQueue()  #우선순위 큐를 이용하여 회의 종료시각을 기준으로 정렬할 것임

for _ in range(N):
  [s,f] = list(map(int, input().split()))
  pq.put((f,[s,f])) #종료시각 f를 기준으로 입력 저장(오름차순-> 종료시각이 빠른 회의일 수록 앞에 배치되게 정렬)
```

```python
# 정렬 확인
print(pq.queue)
```

```python
## 최적해 리스트 생성 및 초기화(greedy한 선택 한번하기 -> pq에서 회의 종료시간이 가장 빠른 가장 앞의 회의를 가져옴)

S = []
S.append(pq.get()[1])


print('S = ', S)
```

```python
## 최적 부분구조 반복
for _ in range(N-1):
  meet = pq.get() #N개의 회의중 첫 회의는 정했으므로 나머지 N-1개의 회의를 회의 종료시간이 빠른 순서대로 차례로 뽑아옴
  if S[-1][1] <= meet[1][0]:  #S[-1][1]은 최적해 리스트에 저장된 가장 최근 회의의 종료시간, meet[1][0]는 현재 겹침 여부 판별하려는 회의의 시작 시간 의미
    S.append(meet[1]) #위 if문을 통과하면 현재 회의 meet[1]이 이전 회의와 겹치지 않는다는 뜻이므로 최적해 리스트에 추가


print('S = ', S)
```

```python
## 출력

print(len(S)) #우리는 최대 넣을 수 있는 회의의 개수를 알고 싶으므로 len(S) 반환
```

```python
## 전체 코드
from queue import PriorityQueue
# import sys
# input = sys.stdin.readline

N = int(input())

pq = PriorityQueue()

for _ in range(N):
  [s,f] = list(map(int, input().split()))
  pq.put((f,[s,f]))

S = []
S.append(pq.get()[1])

for _ in range(N-1):
  meet = pq.get()
  if S[-1][1] <= meet[1][0]:
    S.append(meet[1])

print(len(S))
```

위키피디아 글을 더 읽어보면 Weighted activity selection problem 에 대한 내용이 나온다. 각각의 회의에 가중치가 있을 때 어떻게 회의를 선택해야 하냐는 문제이다(더 중요한 회의가 있을 수 있음). 이런 경우에는 Greedy 알고리즘으로는 풀 수 없으며 Dynamic programming(DP)을 사용해야 한다고 한다. 백준에 이와 관련된 회의실 배정2, 회의실 배정3, ... 과 같이 확장 문제들이 있으니, 나중에 풀어봐도 좋을 듯 싶다.

그럼 마지막으로 DP와 Greedy 알고리즘이 얼마나 효율성 측면에서 차이가 나는지 테스트 해보자. 코드는 GPT로 작성하였다.

```python
## DP로 푸는 코드

N = int(input())
mt = [tuple(map(int, input().split())) for _ in range(N)]

# 끝나는 시간 기준 정렬
mt.sort(key=lambda x: x[1])

dp = [0] * (N + 1)

for i in range(1, N + 1):
    s, f = mt[i - 1]
    # p(i) 찾기: i 이전 회의 중에서 겹치지 않는 가장 마지막 회의
    j = 0
    for k in range(i - 1, 0, -1):
        if mt[k - 1][1] <= s:
            j = k
            break
    # 점화식 적용
    dp[i] = max(dp[i - 1], dp[j] + 1)

print(dp[N])
```

## Greedy vs DP

이제 두 코드를 돌리는데 걸리는 시간을 측정해 비교해보자(greedy vs DP)

```python
import random,time
from queue import PriorityQueue

N=500000  # for문 DP는 O(n^2)이라 너무 크게 하면 너무 느려짐
mt=[(random.randint(0,10**6),0) for _ in range(N)]
mt=[(s,random.randint(s+1,s+1000)) for s,_ in mt]

def greedy(mt):
    pq=PriorityQueue()
    for s,f in mt:pq.put((f,[s,f]))
    S=[pq.get()[1]]
    for _ in range(N-1):
        meet=pq.get()
        if S[-1][1]<=meet[1][0]:S.append(meet[1])
    return len(S)

def dp_for(mt):
    mt.sort(key=lambda x:x[1])
    dp=[0]*(N+1)
    for i,(s,f) in enumerate(mt,1):
        j=0
        for k in range(i-1,0,-1):
            if mt[k-1][1]<=s:
                j=k
                break
        dp[i]=max(dp[i-1],dp[j]+1)
    return dp[N]

t=time.time();print("greedy:",greedy(mt),time.time()-t)
t=time.time();print("dp_for:",dp_for(mt),time.time()-t)
```

greedy 알고리즘보다 DP가 더 시간이 많이 걸리는 것을 확인할 수 있다. 그런데 여기서 DP를 구현하는 방법을 조금 바꾸면 greedy보다 빠르게 연산할 수도 있다고 한다.

DP를 for문으로 구현하는 대신 bisect라는 '정렬된 리스트에서 이분 탐색'을 해주는 파이썬 모듈을 사용하여 구현하면 속도가 아주 빨라진다.

```python
import random,time,bisect
from queue import PriorityQueue

N=500000
mt=[(random.randint(0,10**6),0) for _ in range(N)]
mt=[(s,random.randint(s+1,s+1000)) for s,_ in mt]

def greedy(mt):
    pq=PriorityQueue()
    for s,f in mt:pq.put((f,[s,f]))
    S=[pq.get()[1]]
    for _ in range(N-1):
        meet=pq.get()
        if S[-1][1]<=meet[1][0]:S.append(meet[1])
    return len(S)

def dp(mt):
    mt.sort(key=lambda x:x[1])
    ends=[f for _,f in mt]
    dp=[0]*(N+1)
    for i,(s,f) in enumerate(mt,1):
        j=bisect.bisect_right(ends,s)-1
        dp[i]=max(dp[i-1],dp[j+1]+1)
    return dp[N]

t=time.time();print("greedy:",greedy(mt),time.time()-t)
t=time.time();print("dp:",dp(mt),time.time()-t)
```

greedy도 heapq를 이용하여 구현하면 더 빨라질 수는 있으나 이 문제에 한해서는 bisect를 활용하는 DP가 가장 빠른 듯 하다.

여러가지 정렬방법이나 탐색법을 익혀두는 것이 시간 효율적으로 코드를 짜는데 있어서 유리함을 확인할 수 있다.
